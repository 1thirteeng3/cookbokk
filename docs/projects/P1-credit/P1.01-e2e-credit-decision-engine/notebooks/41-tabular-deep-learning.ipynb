## 2. Dados e pré-processamento

Redes neurais precisam de entradas numéricas bem comportadas. As features numéricas recebem imputação pela mediana do treino, um indicador de ausência e uma transformação por quantis para a normal. As categóricas viram códigos inteiros (0 = ausente ou desconhecida), que o TabM codifica em *one-hot*. Tudo é ajustado só no treino.

## 2. Data and preprocessing

Neural networks need well-behaved numeric inputs. Numeric features get train-median imputation, a missing indicator and a quantile transform to the normal distribution. Categorical features become integer codes (0 = missing or unknown), which TabM one-hot encodes. Everything is fitted on training data only.

```python
torch.manual_seed(SEED)
feats = pd.read_parquet(ARTIFACTS / "features.parquet")
meta = json.loads((ARTIFACTS / "feature_meta.json").read_text())
DATASET = meta["dataset"]
COLS, CAT = meta["model_columns"], meta["categorical"]
NUM = [c for c in COLS if c not in CAT]
y = feats[hc.TARGET].to_numpy()
part = feats["SPLIT"].to_numpy()
tr, va, te = (part == s for s in ("train", "valid", "test"))

Xn = feats[NUM].astype(float)
miss = [c for c in NUM if Xn.loc[tr, c].isna().any()]
filled = Xn.fillna(Xn[tr].median())

def numeric_block(transform=True):
    if not transform:
        return np.hstack([filled.to_numpy(), Xn[miss].isna().to_numpy(float)]).astype(np.float32)
    qt = QuantileTransformer(output_distribution="normal", n_quantiles=min(1000, int(tr.sum())),
                             subsample=10**9, random_state=SEED).fit(filled[tr])
    return np.hstack([qt.transform(filled), Xn[miss].isna().to_numpy(float)]).astype(np.float32)

X_num = numeric_block()
Xc, _ = hc.to_categorical(feats[CAT], meta["categories"])
X_cat = np.stack([Xc[c].cat.codes.to_numpy() + 1 for c in CAT], axis=1).astype(np.int64)
CARDS = [len(meta["categories"][c]) + 1 for c in CAT]
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(X_num.shape, X_cat.shape, DEVICE)
```

## 3. Solução

O TabM produz k previsões por cliente. No treino, a perda é a média das k perdas; na previsão, usamos a média das k probabilidades. A parada antecipada olha o AUC de validação do ensemble.

## 3. Solution

TabM outputs k predictions per client. In training, the loss is the mean of the k losses; at prediction time we average the k probabilities. Early stopping watches the ensemble's validation AUC.

```python
def to_tensors(mask, num=None):
    num = X_num if num is None else num
    return (torch.as_tensor(num[mask], device=DEVICE), torch.as_tensor(X_cat[mask], device=DEVICE),
            torch.as_tensor(y[mask], dtype=torch.float32, device=DEVICE))

@torch.no_grad()
def predict(model, xn, xc, batch=4096):
    model.eval()
    out = [torch.sigmoid(model(xn[i:i + batch], xc[i:i + batch]).squeeze(-1)).mean(1)
           for i in range(0, len(xn), batch)]
    return torch.cat(out).cpu().numpy()

def train_tabm(k=K, num=None, epochs=None, patience=None):
    epochs = epochs or (EPOCHS if RUN_MODE == "full" else 3)
    patience = patience or (PATIENCE if RUN_MODE == "full" else 2)
    xn, xc, yt = to_tensors(tr, num)
    vn, vc, _ = to_tensors(va, num)
    model = tabm.TabM.make(n_num_features=xn.shape[1], cat_cardinalities=CARDS, d_out=1, k=k, d_block=D_BLOCK).to(DEVICE)
    opt = torch.optim.AdamW(model.parameters(), lr=2e-3, weight_decay=3e-4)
    loss_fn = torch.nn.BCEWithLogitsLoss()
    best, best_state, bad_epochs, log = -1.0, None, 0, []
    for epoch in range(epochs):
        model.train()
        for idx in torch.randperm(len(xn), device=DEVICE).split(BATCH_SIZE):
            logits = model(xn[idx], xc[idx]).squeeze(-1)                  # (batch, k)
            loss = loss_fn(logits, yt[idx].unsqueeze(1).expand_as(logits))
            opt.zero_grad(); loss.backward(); opt.step()
        auc = (cr.gini(y[va], predict(model, vn, vc)) + 1) / 2
        log.append(auc)
        if auc > best:
            best, best_state, bad_epochs = auc, copy.deepcopy(model.state_dict()), 0
        else:
            bad_epochs += 1
            if bad_epochs >= patience:
                break
    model.load_state_dict(best_state)
    return model, log

with Timer() as t_fit:
    model, history = train_tabm()
tn, tc, _ = to_tensors(te)
with Timer() as t_pred:
    p_tabm = predict(model, tn, tc)
print(f"{len(history)} epochs, best valid AUC {max(history):.4f}, {t_fit.seconds:.0f}s")
```

### 3.1 Ensemble com o GBM

GBMs e redes erram de jeitos diferentes. Uma média simples dos **postos** (*rank average*) dos dois scores costuma render mais que cada um sozinho, sem exigir calibração conjunta.

### 3.1 Ensemble with the GBM

GBMs and networks make different mistakes. A simple **rank average** of both scores often beats each one alone, without requiring joint calibration.

```python
l3 = pd.read_parquet(ARTIFACTS / "l3_predictions.parquet").set_index(hc.ID)
p_gbm = l3.loc[feats.loc[te, hc.ID], "pd_l3"].to_numpy()
rank = lambda v: pd.Series(v).rank(pct=True).to_numpy()
p_ens = (rank(p_gbm) + rank(p_tabm)) / 2
y_te = y[te]
SCORES = {"LightGBM (L3)": p_gbm, "TabM": p_tabm, "rank average LightGBM + TabM": p_ens}
METRICS = {name: {"gini": cr.bootstrap(cr.gini, y_te, p, n=N_BOOT, seed=SEED),
                  "ks": cr.bootstrap(cr.ks, y_te, p, n=N_BOOT, seed=SEED)} for name, p in SCORES.items()}
pd.DataFrame({n: {"gini": m["gini"]["mean"], "gini_std": m["gini"]["std"], "ks": m["ks"]["mean"]}
              for n, m in METRICS.items()}).T
```

## 4. Validação

A diferença de Gini entre modelos precisa ser maior que o desvio do *bootstrap* para ser levada a sério. A literatura recente (benchmarks de 2024–2025) mostra modelos neurais como o TabM empatando ou vencendo GBMs bem ajustados em parte das bases; em crédito, com poucas features fortes e muitos ausentes, o empate é comum e o ensemble tende a ser o ganho mais estável.

## 4. Validation

The Gini difference between models must exceed the bootstrap deviation to be taken seriously. Recent literature (2024–2025 benchmarks) shows neural models such as TabM tying with or beating well-tuned GBMs on part of the datasets; in credit, with a few strong features and many missing values, a tie is common and the ensemble tends to be the most stable gain.

## 5. Eixos de variação

### 5.1 Tamanho do ensemble interno (k)

Com k = 1, o TabM vira um MLP comum. A diferença mede quanto o ensemble eficiente contribui.

## 5. Variation axes

### 5.1 Internal ensemble size (k)

With k = 1, TabM becomes a plain MLP. The difference measures how much the efficient ensemble contributes.

```python
mlp, _ = train_tabm(k=1)
AXIS_K = {"k=1 (plain MLP)": cr.gini(y_te, predict(mlp, tn, tc)), f"k={K}": METRICS["TabM"]["gini"]["mean"]}
pd.Series(AXIS_K, name="test gini")
```

## 6. Armadilhas demonstradas

### 6.1 Entradas sem transformação

**Sintoma:** treino instável e Gini bem pior. **Causa:** valores como `AMT_CREDIT` (milhões) e `DAYS_BIRTH` (milhares negativos) dominam os gradientes. **Correção:** transformação por quantis (ou padronização robusta) ajustada no treino.

## 6. Pitfalls demonstrated

### 6.1 Untransformed inputs

**Symptom:** unstable training and much worse Gini. **Cause:** values such as `AMT_CREDIT` (millions) and `DAYS_BIRTH` (negative thousands) dominate the gradients. **Fix:** a quantile transform (or robust scaling) fitted on training data.

```python
raw_num = numeric_block(transform=False)
raw_model, _ = train_tabm(num=raw_num, epochs=3, patience=2)
rn, rc, _ = to_tensors(te, raw_num)
pd.Series({"quantile-transformed inputs": METRICS["TabM"]["gini"]["mean"],
           "raw inputs (3 epochs)": cr.gini(y_te, predict(raw_model, rn, rc))}, name="test gini")
```
